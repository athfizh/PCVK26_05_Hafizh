Studi Kasus 1: Label Identitas Dokumentasi Otomatis

```python
def label_dokumentasi(image_path, teks_label="Dokumentasi HMTI Polinema - 2026"):
    img = cv.cvtColor(cv.imread(image_path), cv.COLOR_BGR2RGB)
    tinggi, lebar = img.shape[:2]

    # 1. Tinggi bar adaptif (1/12 dari tinggi total gambar)
    tinggi_bar = tinggi // 12

    # 2. Menggambar bar hitam di dasar gambar
    cv.rectangle(img, pt1=(0, tinggi - tinggi_bar), pt2=(lebar, tinggi), color=(0, 0, 0), thickness=-1)

    # 3. Skala font dan ketebalan dinamis berbasis tinggi gambar
    skala_font = tinggi / 900.0
    tebal_font = max(1, int(tinggi / 300.0))

    # 4. Penentuan posisi origin teks di tengah bar hitam
    posisi_x = lebar // 25
    posisi_y = tinggi - (tinggi_bar // 3)

    cv.putText(img, teks_label, (posisi_x, posisi_y), cv.FONT_HERSHEY_SIMPLEX,
               skala_font, (255, 255, 255), tebal_font, cv.LINE_AA)

    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.title(f"Resolusi {lebar}x{tinggi} px | Tinggi Bar: {tinggi_bar} px")
    plt.axis('off')
    plt.show()

# Pengujian dengan dua citra beda orientasi
label_dokumentasi("/content/drive/MyDrive/COLLEGE/SEM 5/PCVK/CITRAMODUL2/expo'26 HMTIxWRIxITDEC.jpg")
label_dokumentasi("/content/drive/MyDrive/COLLEGE/SEM 5/PCVK/CITRAMODUL2/rapim expo'26.jpg")
```

Studi Kasus 2: Thumbnail Marketplace 1:1 Adaptif

```python
def buat_thumbnail_persegi(image_path, nama_toko):
    img = cv.cvtColor(cv.imread(image_path), cv.COLOR_BGR2RGB)
    tinggi, lebar = img.shape[:2]

    dimensi_maks = max(tinggi, lebar)
    kanvas = np.full((dimensi_maks, dimensi_maks, 3), 255, dtype=np.uint8)

    offset_y = (dimensi_maks - tinggi) // 2
    offset_x = (dimensi_maks - lebar) // 2

    kanvas[offset_y : offset_y + tinggi, offset_x : offset_x + lebar] = img

    skala_font = dimensi_maks / 750.0
    tebal_font = max(1, int(dimensi_maks / 250.0))

    # 1. Hitung panjang dan tinggi kotak teks secara dinamis
    (lebar_teks, tinggi_teks), _ = cv.getTextSize(nama_toko, cv.FONT_HERSHEY_DUPLEX, skala_font, tebal_font)

    # 2. Atur posisi ke Pojok Kanan Bawah (diberi jarak margin 3% dari tepi kanvas)
    posisi_x = dimensi_maks - lebar_teks - int(dimensi_maks * 0.03)
    posisi_y = dimensi_maks - int(dimensi_maks * 0.03)

    posisi_x = (dimensi_maks - lebar_teks) // 2

    cv.putText(kanvas, nama_toko, (posisi_x, posisi_y), cv.FONT_HERSHEY_DUPLEX,
               skala_font, (50, 50, 50), tebal_font, cv.LINE_AA)

    plt.figure(figsize=(6, 6))
    plt.imshow(kanvas)
    plt.title(f"Keluaran Thumbnail 1:1 ({dimensi_maks}x{dimensi_maks} px)")
    plt.axis('off')
    plt.show()

# Panggil fungsinya
buat_thumbnail_persegi("/content/drive/MyDrive/COLLEGE/SEM 5/PCVK/CITRAMODUL2/expo'26 HMTIxWRIxITDEC.jpg", "Expo HMTI 2026")
```
